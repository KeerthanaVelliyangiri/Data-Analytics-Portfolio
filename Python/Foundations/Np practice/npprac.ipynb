concate

```python
b11=np.array([56,78,89])
b22=np.array([65,76,98])

comb=np.concatenate((b11,b22))
print(comb)
```

Split-always equal division 

```python
print(np.split(comb,2))

for arr in np.split(comb,2):
    print(arr)

```

```python
b1=np.array([[56,78,89],[34,45,68]])
b2=np.array([[85,78,90],[89,98,78]])

comb1=np.concatenate((b1,b2),axis=1)
print(comb1)
print("--------------")
comb0=np.concatenate((b1,b2),axis=0)
print(comb0)

```

search

```python
marks=np.array([88,90,30,45,75])
print(np.where(marks==75))
```

```python
mark=np.array([[85,78,90],
      [92,85,98],
      [76,82,80]])

print(np.where(mark==78))
print(mark)
```

```python
mark=np.array([[85,78,90],
      [92,85,98],
      [76,82,80]])

print(np.where(mark<=80)) #gives index
print(mark)
```

```python
mark=np.array([[85,78,90],
      [92,85,98],
      [76,82,80]])

print(mark[np.where(mark>=80)]) #give exact value
print(mark)
```
