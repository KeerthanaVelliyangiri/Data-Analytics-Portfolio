```python
import matplotlib.pyplot as plt
```

```python
dep_avg=df.groupby("Department")["Python"].mean()
dep_avg.plot(kind='bar',color='red')
plt.title('Department wise Python Avg')
plt.ylabel('Avg Python Mark')
plt.xlabel('Department')
plt.Figure(figsize=(20,16))
plt.show()
```
