```python
import numpy as np
```

```python
from sklearn.linear_model import LinearRegression
```

```python
# study_marks
# x is feature
X=np.array([[12],[5],[15],[4]])
# y is target
y=np.array([78, 45, 88, 39])
# choose the model
model=LinearRegression()
# train the model
model.fit(X,y)
# feature value to predict the target value
ans=model.predict([[8]])
print(ans)
# 2d array of feature to convert 1d array bcz corr takes 1d
X1=X.ravel()
# find the dif btw the feature and target - corr
np.corrcoef(X1,y)
```

```python
# attendance
att=np.array([[85],[62],[91],[55]])
y=np.array([78, 45, 88, 39])
model=LinearRegression()
model.fit(att,y)
ans=model.predict([[74]])
print(ans)
X1=att.ravel()
np.corrcoef(X1,y)
```
